```python
import pandas as pd
import numpy as np
```

```python
# Load the dataset using a comma as the delimiter
df = pd.read_csv('/content/marketing_campaign_performance_10000.csv', delimiter=',')

# Display the first 10 rows
display(df.head(10))

# Check the dataset dimensions
print(f"Number of rows: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")
```

```python
df.info()
df.describe()
df.isnull().sum()
df.duplicated().sum()
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
print(df.columns.tolist())
```

```python
# Convert 'startdate' and 'enddate' to datetime objects
df['startdate'] = pd.to_datetime(df['startdate'])
df['enddate'] = pd.to_datetime(df['enddate'])

# Display the updated data types to confirm
print("Updated DataFrame Info after datetime conversion:")
df.info()
```

```python
# Convert 'channel' to a categorical data type
df['channel'] = df['channel'].astype('category')

# Display the updated data types to confirm
print("Updated DataFrame Info after category conversion:")
df.info()
```

```python
# Check for empty strings in object/category columns
(df.select_dtypes(include=['object', 'category']) == '').sum()
```

## Handling Missing Values

The dataset was inspected for missing values using `df.isnull().sum()`.

No missing values were found across any of the 11 columns.

Since the dataset is complete, no imputation or row removal was required. All 10,000 records were retained for subsequent analysis.

Maintaining the full dataset ensures that later calculations—such as campaign revenue, conversion rates, ROI comparisons, and channel performance—are based on complete information without introducing bias through data imputation.

```python
# Check for duplicate rows
duplicate_count = df.duplicated().sum()

print(f"Number of duplicate rows: {duplicate_count}")
```

## Duplicate Record Check

The dataset was examined for duplicate records using `df.duplicated().sum()`.

No duplicate rows were identified.

Since duplicate campaign records can artificially inflate marketing KPIs such as impressions, clicks, conversions, revenue, and advertising spend, verifying their absence helps ensure the accuracy of subsequent analyses.

No records were removed during this step.

```python
numeric_cols = [
    'impressions',
    'clicks',
    'leads',
    'conversions',
    'cost_usd',
    'revenue_usd',
    'roi'
]

for col in numeric_cols:
    count = (df[col] < 0).sum()
    print(f"{col}: {count} negative values")

for col in numeric_cols:
    count = (df[col] == 0).sum()
    print(f"{col}: {count} zero values")

(df['clicks'] > df['impressions']).sum()

(df['leads'] > df['clicks']).sum()

(df['conversions'] > df['leads']).sum()

(df['enddate'] < df['startdate']).sum()

df['campaign_duration'] = (
    df['enddate'] - df['startdate']
).dt.days

df['campaign_duration'].describe()

(df['revenue_usd'] < df['cost_usd']).sum()

df['roi'].describe()

```

## Numerical Data Validation

The numerical variables were validated to ensure that campaign metrics were logically consistent and suitable for analysis.

The validation included:

- Checking for negative values.
- Checking for zero values.
- Verifying the marketing funnel hierarchy:
  - Impressions ≥ Clicks ≥ Leads ≥ Conversions.
- Confirming that campaign end dates occurred after campaign start dates.
- Reviewing campaign duration.
- Examining campaigns where marketing cost exceeded generated revenue.
- Inspecting the distribution of ROI values.

These validation checks help ensure that subsequent KPI calculations and campaign performance analyses are based on internally consistent data.

```python
df.select_dtypes(include=['object', 'category']).columns

for col in ['campaignid', 'channel']:
    print(f"\n{col.upper()}")
    print(df[col].unique())

print(df['channel'].value_counts())

df['channel'] = (
    df['channel']
      .str.strip()
      .str.title()
)

print(df['channel'].value_counts())
```

## Standardization of Categorical Variables

Categorical variables were reviewed to identify inconsistent naming conventions that could affect aggregation and reporting.

The `channel` column was inspected for variations in capitalization, spacing, and formatting.

After inspection, category labels were standardized where necessary to ensure that each marketing channel was represented consistently throughout the analysis.

The `campaignid` column was retained without modification, as each value serves as a unique campaign identifier.
